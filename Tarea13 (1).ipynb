# **Ejercicio 13**

Escribe un programa en Python que muestra al usario el menu siguiente:


1.   Saludar.
2.   Mostrar un mensaje.
3.   Salir

Mientras el usario no presione la opción **3**, el programa debe seguir mandando el menu. Si el usario presiona **1**, el programa pide el nombre al usario y muetra a pantalla un saludo con su nombre. Si el usario presiona **2**, el programa debe mandar a panta lla un mensaje personalizado por ustedes.

Cada una de las lienas de código escritas deben estar acompañadas de un comentario que explique su uso.



```python
#Escribe tu código.
# Inicio del programa: Usamos un ciclo while para repetir el menu hasta que el usuario elija salir
opcion = 0 # Inicializo la variable 'opcion' para poder entrar al ciclo

while opcion != 3: # Mientras la opcion no sea 3, el menu se seguira mostrando
  print("1. Saludar")  # Muestro la opcion 1 del menu
  print("2. Mostrar un mensaje")  # Muestro la opcion 2 del menu
  print("3. Salir")  # Opcion para salir del programa

  opcion =int(input("Elige una opcion:"))  # Pido al usuario que ingrese una opcion

  if opcion == 1:  # Si elige la opcion 1, entonces vamos a saludar
    nombre = input("¿Cual es tu nombre?:")  # Pido el nombre del usuario
    print("¡Hola", nombre + "! Encantado de saludarte!")  # Mando un saludo usando su nombre

  elif opcion == 2:  # Si elige la opcion 2, mostramos un mensaje propio
    print("Este es un mensaje especial: ¡Sigue practicando y llegaras lejos!")  # Mensaje personalizado

  elif opcion == 3:  # Si elige la opcion 3, salimos del programa
    print("¡Hasta luego!")  # Mensaje de despedida

  else: # Si escribe un numero que no esta en el menu
    print("Opcion invalida. Por favor, elige una opcion valida.")  # Mensaje de error
```
